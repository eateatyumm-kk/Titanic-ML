```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
train_data = pd.read_csv('../data/train.csv')
test_data = pd.read_csv('../data/test.csv')
```

```python
train_test = [train_data, test_data]

for dataset in train_test:
    dataset['Title'] = (dataset['Name']
    .str.extract(' ([A-Za-z]+)\.', expand=False)
    .replace(['Lady', 'Countess','Capt', 'Col', 'Don', 'Dr', 'Major', 'Rev', 'Sir', 'Jonkheer', 'Dona'], "Rare")
    .replace({'Mlle': 'Miss', 'Ms': 'Miss', 'Mme': 'Mrs'}))
```

```python
train_data['Age'] = train_data['Age'].fillna(train_data.groupby('Title')['Age'].transform('median'))
test_data['Age'] = test_data['Age'].fillna(test_data.groupby('Title')['Age'].transform('median'))

train_data = pd.get_dummies(train_data, columns=["Title"],dtype=int)
test_data = pd.get_dummies(test_data, columns=["Title"],dtype=int)
train_data.head()
```

```python
train_data['Sex'] = train_data['Sex'].replace({'male': 0, 'female': 1})
test_data['Sex'] = test_data['Sex'].replace({'male': 0, 'female': 1})
train_data.head(15)
```

```python
train_data = train_data.drop('Name', axis=1)
test_data = test_data.drop('Name', axis=1)
train_data.head()
```

```python
drop_cols = ['Ticket', 'Cabin', 'PassengerId']
train_data = train_data.drop(drop_cols, axis=1)
test_data = test_data.drop(drop_cols, axis=1)

train_data.head()
```

```python
train_data['Embarked'] = train_data['Embarked'].fillna('S')
```

```python
test_data['Fare'] = test_data['Fare'].fillna(test_data.groupby('Pclass')['Fare'].transform('median'))
test_data.isnull().sum()
```

```python
train_test = [train_data, test_data]

for dataset in train_test:
    dataset['Embarked'] = dataset['Embarked'].map({'S': 0, 'C': 1, 'Q': 2})

#SibSP = number of siblings / spouses aboard the Titanic, #Parch = number of parents / children aboard the Titanic

train_data['FamilySize'] = train_data['SibSp'] + train_data['Parch'] + 1
test_data['FamilySize'] = test_data['SibSp'] + test_data['Parch'] + 1

train_data.head(10)
```

```python
train_data = train_data.drop(['SibSp', 'Parch'], axis=1)
test_data = test_data.drop(['SibSp', 'Parch'], axis=1)

train_data.head()
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")

class Net(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(11, 16)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(16, 8)
        self.fc3 = nn.Linear(8, 1) # output layer

    def forward(self, x):
        x = self.relu(self.fc1(x))
        x = self.relu(self.fc2(x))
        out = self.fc3(x)
        return out

model = Net().to(device)
loss_fn = nn.BCEWithLogitsLoss() #there are different type of loss functions BCEW is for 0 and 1 output -> study different type of lossfn 
optimizer = optim.Adam(model.parameters(), lr=0.005) #there are different type of gradient decent -> study different type of opt function

X_numpy = train_data.drop('Survived', axis=1).values.astype(np.float32)
y_numpy = train_data['Survived'].values.astype(np.float32)

X_train_raw, X_val_raw, y_train_raw, y_val_raw = train_test_split(
    X_numpy, y_numpy, test_size=0.2, random_state=42, stratify=y_numpy
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_raw)
X_val_scaled = scaler.transform(X_val_raw)

X_train = torch.tensor(X_train_scaled, dtype=torch.float32, device=device) #we don't put requires_grad=True here because we don't need gradient of input against y
y_train = torch.tensor(y_train_raw, dtype=torch.float32, device=device).unsqueeze(1)

X_val = torch.tensor(X_val_scaled, dtype=torch.float32, device=device)
y_val = torch.tensor(y_val_raw, dtype=torch.float32, device=device).unsqueeze(1)

# 6. Training Loop with Early Stopping
max_epochs = 1000
patience = 30           # Stop if val loss doesn't improve for 30 epochs
best_val_loss = float('inf')
patience_counter = 0

for epoch in range(max_epochs):
    # Training
    model.train()
    optimizer.zero_grad()
    train_pred = model(X_train)
    train_loss = loss_fn(train_pred, y_train)
    train_loss.backward()
    optimizer.step()

    # Validation
    model.eval()
    with torch.no_grad():
        val_pred = model(X_val)
        val_loss = loss_fn(val_pred, y_val).item()

    # Early Stopping check
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        patience_counter = 0  # Reset
    else:
        patience_counter += 1

    if patience_counter >= patience:
        print(f" Stopped early at epoch {epoch + 1}! Best Val Loss: {best_val_loss:.4f}")
        break

    if (epoch + 1) % 50 == 0:
        print(f"Epoch [{epoch + 1}/{max_epochs}] | Train Loss: {train_loss.item():.4f} | Val Loss: {val_loss:.4f}")
```

```python
model.eval()
with torch.no_grad():
    # 1. Get raw output logits for validation data
    val_logits = model(X_val)
    
    # 2. Convert logits to probabilities (0.0 to 1.0)
    val_probs = torch.sigmoid(val_logits)
    
    # 3. Convert probabilities to 0 or 1 predictions (threshold at 0.5)
    val_preds = (val_probs >= 0.5).float()
    
    # 4. Compare predictions with actual y_val targets
    correct = (val_preds == y_val).sum().item()
    total = y_val.size(0)
    accuracy = (correct / total) * 100

print(f" Validation Accuracy: {accuracy:.2f}%")
```
